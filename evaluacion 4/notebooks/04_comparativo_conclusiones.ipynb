# 04 · Comparación y conclusiones

## 1. ¿Qué problema resuelve cada componente?

El supervisado utiliza etiquetas para predecir si hay una falla APS y reducir el costo de los errores. El no supervisado busca grupos de funcionamiento y registros atípicos sin usar la etiqueta durante el entrenamiento.

## 2. ¿Qué algoritmos se eligieron y por qué?

Se compararon regresión logística, Random Forest y XGBoost: la primera es una línea base lineal y los otros dos captan relaciones no lineales. XGBoost fue elegido por su costo en validación. En el no supervisado se usaron K-Means, que permite describir grupos mediante distancias, e Isolation Forest, que identifica registros fáciles de aislar. K-Means favorece grupos compactos; Isolation Forest depende de la proporción de anomalías elegida.

## 3. ¿Qué dificultades aparecieron?

Hay muchos valores faltantes, variables redundantes y distribuciones asimétricas; además, las fallas son escasas y sus errores tienen costos distintos. Se usaron imputación con indicadores, transformaciones y validación en entrenamiento. Para reducir el costo computacional del análisis no supervisado se tomó una muestra aleatoria de 10.000 registros. Sus resultados pueden variar con otra muestra y no representan necesariamente todos los perfiles.

## 4. ¿El supervisado identificó los casos de interés?

En el test oficial, XGBoost con umbral 0,005 identificó 365 de 375 fallas: recall 0,973, precisión 0,465, F1 0,629 y PR-AUC 0,936. Hubo 420 falsos positivos y 10 falsos negativos, para un costo de 9.200 con la regla de la actividad. Son resultados del test documentados en el notebook 02, no una garantía de desempeño en una flota nueva.

## 5. ¿El no supervisado descubrió estructuras?

En la muestra, K-Means encontró cuatro grupos. El grupo 3 concentró 122 de las 170 fallas y presentó medianas mayores en `aa_000` y `ck_000` que los demás. Isolation Forest marcó 200 registros, 52 con falla. Esto revela concentraciones útiles para investigar, aunque no demuestra que exista una causa común y depende de la muestra y los parámetros.

## 6. ¿Cómo se relacionan los grupos con las etiquetas?

La tasa de fallas fue 0,21 % en el grupo 0, 0,74 % en el 1, 3,45 % en el 2 y 11,41 % en el 3. La etiqueta se usó solo después de entrenar. Los grupos no equivalen a clases: el grupo 3 también contiene muchos registros sin falla y el grupo 2 tiene solo 87 registros, por lo que su tasa es inestable.

## 7. ¿Qué se conserva o se pierde con PCA?

Los 53 componentes usados conservan el 90,3 % de la varianza de las 141 variables preparadas en esta muestra. Se pierde el resto de la variación y parte de la interpretación directa de sensores; una señal de falla puede estar en una dirección de baja varianza. La gráfica de dos dimensiones conserva todavía menos información que los 53 componentes usados para entrenar.

## 8. ¿Qué enfoque recomendaríamos?

Para priorizar revisiones, recomendaríamos el modelo supervisado con su umbral de costo, una vez validado con datos recientes y costos reales. El análisis no supervisado sirve como complemento para investigar perfiles y comportamientos nuevos, no como diagnóstico definitivo. Las variables están anonimizadas y se necesita validación con personal de mantenimiento antes de actuar.

## Conclusión

Los dos enfoques responden preguntas distintas: el supervisado detecta la falla conocida y el no supervisado ayuda a explorar perfiles. La concentración observada de fallas entre anomalías justifica investigarlas, pero no permite afirmar que todas sean fallas ni que se identifiquen todas.
